```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
import glob
import pickle as pkl
import sim_name_remap as map
import plotter_config as pc
```

```python

SAVEDIR = '/glade/u/home/wsmith/NudgeComparison_scripts/vars/'

sims = ['2022_E5-strong_h0','2022_M2-strong_h0','2022_E5-weak_h0','2022_M2-weak_h0','2022_FreeRun_h0']   # TROP_P did not make it into the h0 files....
vars = ['CO','T','CO_CHML','NOx','O3']#'E90','N2O','SF6','CFC12']
date0 = '2022-05'   # For h0 files
date1 = '2022-09'

#sims = ['2022_E5-strong','2022_M2-strong','2022_E5-weak','2022_M2-weak']   # TROP_P did not make it into the h0 files....
#vars = ['TROP_P']
#date0 = '2022-05-01'   # For h0 files
#date1 = '2022-09-30'
```

```python

for sim in sims: 

    FILES = sorted(glob.glob(map.remap[sim]))
    FILES0 = [item for item in FILES if date0 in item][0]
    find0 = list(FILES).index(FILES0)
    FILES1 = [item for item in FILES if date1 in item][0]
    find1 = list(FILES).index(FILES1)
    FILES = FILES[find0:find1+1]

    for var_name in vars:

        var_str = pc.var_info[var_name]['vars']
        
        print(sim)
        print(var_str)

        SAVEFILE = SAVEDIR + sim + '_' + var_name + '_avgXsect_' + date0 + '_' + date1 + '.pkl'
        
        ds = xr.open_mfdataset(
            FILES,
            combine="by_coords",
            parallel=True
        )
        
        var = ds[var_str[0]]
        if len(var_str) > 1: 
            for v in range(len(var_str)-1):
                curr_var = ds[var_str[v+1]]
                var = var + curr_var
        
        var_mean = var.mean(dim="time")
        var_zonalmean = var_mean.mean(dim="lon")       
        
        with open(SAVEFILE, 'wb') as f:   
            pkl.dump(var_zonalmean, f)  
```
